```python
#1.Bayesian Reasoning:
print("BAYESIAN REASONING")
print("1. Medical Diagnosis")
print("2. Autonomous Vehicle")
print("3. Smart Manufacturing")

choice = int(input("Enter your choice: "))

if choice == 1:
    # Medical Diagnosis
    prior = 0.10
    sensitivity = 0.90
    false_positive = 0.05

    numerator = sensitivity * prior
    denominator = numerator + false_positive * (1 - prior)

    result = numerator / denominator

    print("Probability of disease:", round(result * 100, 2), "%")

elif choice == 2:
    # Autonomous Vehicle
    pedestrian = 0.70
    vehicle = 0.20
    object = 0.10

    print("Pedestrian probability:", pedestrian * 100, "%")
    print("Vehicle probability:", vehicle * 100, "%")
    print("Stationary object probability:", object * 100, "%")

    print("Detected object: Pedestrian")

elif choice == 3:
    # Smart Manufacturing
    temperature = int(input("Enter temperature level (1=Normal, 2=High): "))
    vibration = int(input("Enter vibration level (1=Normal, 2=High): "))

    if temperature == 2 and vibration == 2:
        probability = 85
    elif vibration == 2:
        probability = 60
    elif temperature == 2:
        probability = 30
    else:
        probability = 5

    print("Machine failure probability:", probability, "%")

    if probability >= 60:
        print("Maintenance Required")
    else:
        print("Machine Working Normally")

else:
    print("Invalid choice")
```

```python
#2.Fuzzy Inferencing System:
print("FUZZY INFERENCING SYSTEM")
print("1. Medical Diagnosis")
print("2. Autonomous Vehicle")
print("3. Smart Manufacturing")

choice = int(input("Enter your choice: "))

if choice == 1:
    # Medical Diagnosis
    temp = float(input("Enter high temperature membership (0-1): "))
    heart = float(input("Enter high heart rate membership (0-1): "))

    risk = min(temp, heart)

    if risk >= 0.7:
        print("Disease Risk: HIGH")
    elif risk >= 0.3:
        print("Disease Risk: MEDIUM")
    else:
        print("Disease Risk: LOW")

elif choice == 2:
    # Autonomous Vehicle
    distance = float(input("Enter near distance membership (0-1): "))
    speed = float(input("Enter fast speed membership (0-1): "))

    braking = max(distance, min(distance, speed))

    if braking >= 0.7:
        print("Braking: HIGH")
    elif braking >= 0.3:
        print("Braking: MEDIUM")
    else:
        print("Braking: LOW")

elif choice == 3:
    # Smart Manufacturing
    temp = float(input("Enter high temperature membership (0-1): "))
    vibration = float(input("Enter high vibration membership (0-1): "))

    danger = min(temp, vibration)

    if danger >= 0.7:
        print("Machine Speed: SLOW")
    elif danger >= 0.3:
        print("Machine Speed: NORMAL")
    else:
        print("Machine Speed: FAST")

else:
    print("Invalid choice")
```
